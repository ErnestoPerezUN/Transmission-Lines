```python
import numpy as np
```

```python
mu_0=4*np.pi*1e-7  # Permeability of free space in T*m/A
eps_0=8.854187817e-12  # Permittivity of free space in F/m

r=.001
s=1
mu_r=1  # Relative permeability of the material
eps_r=1  # Relative permittivity of the material
```

```python
mu=mu_0*mu_r  # Permeability of the material in T*m/A
eps=eps_0*eps_r  # Permittivity of the material in F/m
l=mu/(2*np.pi)*np.log((s-r)**2/r**2)  # Inductance of a solenoid in H
l_aprox=mu/(np.pi)*np.log(s/r)  # Approximate inductance of a solenoid in H
c=eps*np.pi/np.log((s-r)/r)  # Capacitance of a solenoid in F
c_aprox=eps*np.pi/np.log(s/r) 
err=np.abs(l_aprox-l)/l  # Error in the approximation
```

```python
display("Inductance : {:.5e} H".format(l))
display("Approximate : {:.5e} H".format(l_aprox))
display("Error in approximation L: {:.5%}".format(err))
display("Capacitance : {:.5e} F".format(c))
display("Approximate : {:.5e} F".format(c_aprox))
display("Error in approximation C: {:.5%}".format(np.abs(c_aprox-c)/c))
```
